**Imports**

```python
import pyspark.sql.functions as F
```

```python
bike_df=spark.read.table("rental.bike_rental_bronze.bike")
```

```python
display(bike_df)
```

**Product category Standardlization**

```python
bike_df = bike_df.withColumn(
    "category",
    F.when(F.lower(F.col("category")).rlike("road"), "Road")
     .when(F.lower(F.col("category")).rlike("mountain"), "Mountain")
     .when(F.lower(F.col("category")).rlike("city"), "City")
     .when(F.lower(F.col("category")).rlike("hybrid"), "Hybrid")
     .when(F.lower(F.col("category")).rlike("electric"), "Electric")
     .otherwise("Other")
)
```

**Status Normalization**

```python
bike_df = bike_df.withColumn(
    "status",
    F.when(F.lower(F.col("status")).isin("available", "rented", "out of service"), F.col("status"))
     .otherwise("Other")
)
```

**utilization_flag**

```python
bike_df = bike_df.withColumn(
    "utilization_flag", F.when(F.col("status") == "rented", "Y").otherwise("N")
)
```

```python
bike_df.write.mode("overwrite").saveAsTable("rental.bike_rental_silver.bike")
```
