```python
%sql
CREATE TABLE IF NOT EXISTS rental.bike_rental_gold.dim_date (
    date_id INT,
    full_date DATE,
    day INT,
    month INT,
    quarter STRING,
    year INT,
    weekday STRING,
    week_number INT
);
```

```python
%sql
INSERT INTO rental.bike_rental_gold.dim_date
(
    date_id,
    full_date,
    day,
    month,
    quarter,
    year,
    weekday,
    week_number
)
SELECT
    CAST(date_format(d, 'yyyyMMdd') AS INT) AS date_id,
    d AS full_date,
    day(d) AS day,
    month(d) AS month,
    CONCAT('Q', quarter(d)) AS quarter,
    year(d) AS year,
    date_format(d, 'EEEE') AS weekday,
    weekofyear(d) AS week_number
FROM (
    SELECT explode(
        sequence(
            DATE '2022-10-01',
            DATE '2023-12-31',
            INTERVAL 1 DAY
        )
    ) AS d
);
```
