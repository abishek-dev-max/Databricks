**Imports**

```python
import pyspark.sql.functions as F
```

**Reading The Table**

```python
dim_membership= spark.read.table("rental.bike_rental_silver.membership")
```

```python
dim_membership = dim_membership.withColumn("membership_name",F.lit("Daily")).withColumn("price",F.lit(500))
```

```python
dim_membership = dim_membership.select("id","membership_type_id","membership_name","price")
```

**Writing to Gold**

```python
dim_membership.write.mode("overwrite").option("overwriteSchema", True).saveAsTable("rental.bike_rental_gold.dim_membership_type")
```
