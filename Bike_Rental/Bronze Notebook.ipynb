**Customer**

```python
%sql
create or replace table rental.bike_rental.customer
(
	id		int primary key,
	name	varchar(30),
	email	varchar(50)
);
```

**Bike Model**

```python
%sql
create or replace table rental.bike_rental.bike
(
	id					int primary key,
	model				varchar(50),
	category			varchar(50),
	price_per_hour		decimal,
	price_per_day		decimal,
	status				varchar(20)
);
```

**Rental**

```python
%sql
create or replace table rental.bike_rental.taken_rent
(
	id					int primary key,
	customer_id			int ,
	bike_id				int ,
	start_timestamp		timestamp,
	duration			int,
	total_paid			decimal
);
```

**Membership Type**

```python
%sql
create or replace table rental.bike_rental.membership_type
(
	id				int primary key,
	name			varchar(50),
	description		varchar(500),
	price			decimal
);
```

```python
%sql
create or replace table rental.bike_rental.membership
(
	id					int primary key,
	membership_type_id	int ,
	customer_id			int ,
	start_date			date,
	end_date			date,
	total_paid			decimal
);
```

```python
df=spark.read.table("rental.bike_rental_bronze.bike")
df=df.withColumn("Ingestion_Time",F.current_timestamp())
```

```python
df.write.option("overwriteSchema", "true").mode("overwrite").saveAsTable("rental.bike_rental_bronze.bike")
```
